# Creating labels and annotations

```python
import numpy as np
import pandas as pd
from pandas import DataFrame, Series

import matplotlib.pyplot as plt
from pylab import rcParams
import seaborn as sb
```

```python
%matplotlib inline
rcParams['figure.figsize'] = 12,4
sb.set_style('whitegrid')
```

## Labelling Plot Feature

### The Functional method

```python
x = range(1, 10)
y = [3, 2, 5, 2, 5, 8, 2, 7, 4]

plt.bar(x,y)
plt.xlabel("x-axis label")
plt.ylabel("Y-axis label")
```

```python
z = [1, 2, 3, 4, 0.5]
veh_type = ['bicycle', 'motorbike', 'car', 'van', 'truck']
plt.pie(z, labels=veh_type)
plt.show()
```

### The Object oriented method

```python
address = 'D:\Tutorials\AI - Engineer\Data_Science_and_ML\data\mtcars.csv'

cars = pd.read_csv(address)
cars.columns = ['car_names','mpg','cyl','disp', 'hp', 'drat', 'wt', 'qsec', 'vs', 'am', 'gear', 'carb']

mpg = cars.mpg

fig = plt.figure()

ax = fig.add_axes([.1, .1, 1, 1])

mpg.plot(ax=ax)

ax.set_xticks(range(len(cars.car_names)))
ax.set_xticklabels(cars.car_names, rotation=60, fontsize='medium')
ax.set_title("Miles per Gallon of cars in mtcars")

ax.set_xlabel('car names')
ax.set_ylabel('miles/gallon')
```

## Adding a legend to your plot

### The Functional Method

```python
plt.pie(z)
plt.legend(veh_type, loc='best')
plt.show()
```

### The Object Oriented Method

```python
fig = plt.figure()

ax = fig.add_axes([.1, .1, 1, 1])

mpg.plot(ax=ax)

ax.set_xticks(range(len(cars.car_names)))
ax.set_xticklabels(cars.car_names, rotation=60, fontsize='medium')
ax.set_title("Miles per Gallon of cars in mtcars")

ax.set_xlabel('car names')
ax.set_ylabel('miles/gallon')

ax.legend(loc='best')
```

```python
fig = plt.figure()

ax = fig.add_axes([.1, .1, 1, 1])

mpg.plot(ax=ax)

ax.set_xticks(range(len(cars.car_names)))
ax.set_xticklabels(cars.car_names, rotation=60, fontsize='medium')
ax.set_title("Miles per Gallon of cars in mtcars")

ax.set_xlabel('car names')
ax.set_ylabel('miles/gallon')

ax.set_ylim([0, 45])

ax.annotate('Toyota Corolla', xy=(19,33.9), xytext=(21,35),arrowprops=dict(facecolor='black', shrink=0.05))

ax.legend(loc='best')
```
